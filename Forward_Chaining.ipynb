```python

import re


class Rule:
    def __init__(self, premises, conclusion):
        self.premises = premises
        self.conclusion = conclusion

    def __repr__(self):
        premises_str = " ^ ".join(
            f"{p}({','.join(args)})" for p, args in self.premises
        )
        pred, args = self.conclusion
        return f"{premises_str} => {pred}({','.join(args)})"


# --- Core Helper Functions ---

def is_variable(x):
    return isinstance(x, str) and len(x) > 0 and x[0].islower()


def standardize_variables(rule, counter):
    subst = {}
    new_premises = []

    for pred, args in rule.premises + [rule.conclusion]:
        for arg in args:
            if is_variable(arg) and arg not in subst:
                subst[arg] = f"{arg}{counter}"
                counter += 1

    for pred, args in rule.premises:
        new_premises.append(
            (pred, [subst.get(a, a) for a in args])
        )

    c_pred, c_args = rule.conclusion
    new_conclusion = (
        c_pred,
        [subst.get(a, a) for a in c_args]
    )

    return Rule(new_premises, new_conclusion), counter


def subst_predicate(theta, predicate):
    pred_name, args = predicate
    new_args = []

    for arg in args:
        val = arg

        while is_variable(val) and val in theta:
            val = theta[val]

        new_args.append(val)

    return (pred_name, new_args)


# --- Unification Algorithm ---

def unify(x, y, theta=None):
    if theta is None:
        theta = {}

    if theta is False:
        return False

    if x == y:
        return theta

    elif is_variable(x):
        return unify_var(x, y, theta)

    elif is_variable(y):
        return unify_var(y, x, theta)

    elif isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return False

        return unify(x[1], y[1], theta)

    elif isinstance(x, list) and isinstance(y, list):

        if len(x) != len(y):
            return False

        if not x:
            return theta

        new_theta = unify(x[0], y[0], theta)

        return unify(x[1:], y[1:], new_theta)

    else:
        return False


def unify_var(var, val, theta):
    if var in theta:
        return unify(theta[var], val, theta)

    elif is_variable(val) and val in theta:
        return unify(var, theta[val], theta)

    new_theta = theta.copy()
    new_theta[var] = val

    return new_theta


# --- Forward Chaining Logic ---

def fol_fc_ask(kb_rules, kb_facts, query):
    current_facts = list(kb_facts)
    var_counter = 1

    print("\n--- Starting Forward Chaining Inference Trace ---")

    # Check if query already exists
    for fact in current_facts:
        phi = unify(fact, query)

        if phi is not False:
            print(f"Goal matched initially with substitution: {phi}")
            return phi

    while True:
        new_sentences = []

        for rule in kb_rules:
            std_rule, var_counter = standardize_variables(
                rule, var_counter
            )

            # Match premises recursively against facts
            def match_premises(premises, theta):

                if not premises:
                    yield theta
                    return

                first_premise = premises[0]
                rest_premises = premises[1:]

                subst_premise = subst_predicate(
                    theta, first_premise
                )

                for fact in current_facts:
                    new_theta = unify(
                        subst_premise, fact, theta
                    )

                    if new_theta is not False:
                        yield from match_premises(
                            rest_premises, new_theta
                        )

            for theta in match_premises(std_rule.premises, {}):

                q_prime = subst_predicate(
                    theta, std_rule.conclusion
                )

                # Check for duplicate facts
                already_exists = any(
                    unify(q_prime, f, {}) is not False
                    for f in current_facts + new_sentences
                )

                if not already_exists:
                    new_sentences.append(q_prime)

                    print(
                        f"Inferred: {q_prime[0]}({','.join(q_prime[1])})"
                    )

                    phi = unify(q_prime, query)

                    if phi is not False:
                        print(
                            f"Success! Query reached with substitution: {phi}"
                        )
                        return phi

        if not new_sentences:
            print("No new facts can be inferred. Search complete.")
            break

        current_facts.extend(new_sentences)

    return False


# --- Parsers ---

def parse_predicate(text):
    match = re.fullmatch(
        r"\s*(\w+)\s*\(\s*([^()]*)\s*\)\s*",
        text
    )

    if not match:
        raise ValueError(f"Invalid format: {text}")

    pred = match.group(1)

    args_str = match.group(2)

    args = [
        arg.strip()
        for arg in args_str.split(",")
    ]

    if not args_str.strip() or any(not arg for arg in args):
        raise ValueError("Predicate must contain valid arguments.")

    return (pred, args)


def parse_rule(text):
    if "=>" not in text:
        raise ValueError("Rules must contain '=>' notation.")

    left, right = text.split("=>", 1)

    premises = [
        parse_predicate(p)
        for p in left.split("^")
        if p.strip()
    ]

    if not premises:
        raise ValueError("Rule must contain at least one premise.")

    return Rule(premises, parse_predicate(right))


# --- Main Program Entry ---

def main():

    print("===============================================================")
    print("       First-Order Logic Forward Chaining Engine")
    print("===============================================================")

    print("Rules formatting:")
    print("- Use lowercase names for variables: x, y, z")
    print("- Use uppercase names for constants: Robert, CountryA")
    print("- Connect conditions using '^'")
    print("- Use '=>' for deductions\n")

    facts = []

    print("--- Enter Initial Facts (Press Enter to finish) ---")

    while True:
        raw_fact = input("Fact: ").strip()

        if not raw_fact:
            break

        try:
            facts.append(parse_predicate(raw_fact))
        except Exception as e:
            print(f"Error parsing fact: {e}")

    rules = []

    print("\n--- Enter Rules (Press Enter to finish) ---")

    while True:
        raw_rule = input("Rule: ").strip()

        if not raw_rule:
            break

        try:
            rules.append(parse_rule(raw_rule))
        except Exception as e:
            print(f"Error parsing rule: {e}")

    print("\n--- Enter Target Query ---")

    while True:
        raw_query = input("Query: ").strip()

        if raw_query:
            try:
                query = parse_predicate(raw_query)
                break
            except Exception as e:
                print(f"Error parsing query: {e}")

    result = fol_fc_ask(rules, facts, query)

    print("\n--- Final Summary ---")
    print(f"Result: {result is not False}")


if __name__ == "__main__":
    main()
```
