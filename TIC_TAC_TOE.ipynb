```python
matrix = [[1,2,3],[4,5,6],[7, 8, 9]]
def print_board():
    for row in matrix:
        print(f" {row[0]} | {row[1]} | {row[2]} ")
    print()
def is_valid_move(pos):
    for r in range(3):
        for c in range(3):
            if matrix[r][c] == pos:
                return r, c
    return None
def check_win(p):
    for i in range(3):
        if all(matrix[i][j] == p for j in range(3)) or all(matrix[j][i] == p for j in range(3)):
            return True
    if matrix[0][0] == matrix[1][1] == matrix[2][2] == p:
        return True
    if matrix[0][2] == matrix[1][1] == matrix[2][0] == p:
        return True
    return False
def play():
    player = "X"
    for turn in range(9):
        print_board()
        try:
            pos = int(input(f"Player {player}, enter position (1-9): "))
            coords = is_valid_move(pos)

            if coords:
                r, c = coords
                matrix[r][c] = player

                if check_win(player):
                    print_board()
                    print(f"Player {player} wins!")
                    return

                player = "O" if player == "X" else "X"
            else:
                print("Spot taken or invalid! Try again.")
        except ValueError:
            print("Please enter a number.")
    print_board()
    print("It's a tie!")
play()
```
