```python

from itertools import product

# Implication function
def implies(p, q):
    return (not p) or q


# Equivalence function
def iff(p, q):
    return p == q


# Evaluate all propositions
def evaluate(R, S):

    # (i) a: (R -> S) AND (S -> R)
    a = implies(R, S) and implies(S, R)

    # (ii) b: R <-> S
    b = iff(R, S)

    # (iii) c: NOT R OR S
    c = (not R) or S

    # (iv) d: NOT (R <-> S)
    d = not iff(R, S)

    # (v) e: (R AND S) OR (NOT R AND NOT S)
    e = (R and S) or ((not R) and (not S))

    return a, b, c, d, e


# Display truth table
def truth_table():

    print("\nTRUTH TABLE")
    print("-" * 85)

    print(f"{'R':^5} {'S':^5} {'a':^8} {'b':^8} "
          f"{'c':^8} {'d':^8} {'e':^8}")

    print("-" * 85)

    # Generate all possible truth assignments
    for R, S in product([False, True], repeat=2):

        a, b, c, d, e = evaluate(R, S)

        def tf(value):
            return 'T' if value else 'F'

        print(f"{tf(R):^5} {tf(S):^5} {tf(a):^8} "
              f"{tf(b):^8} {tf(c):^8} "
              f"{tf(d):^8} {tf(e):^8}")

    print("-" * 85)


# Check propositional entailment
def entails(target):

    for R, S in product([False, True], repeat=2):

        a, b, c, d, e = evaluate(R, S)

        # If a is true and target is false,
        # entailment does not hold
        if a and not target(R, S):
            return False

    return True


# Main program
truth_table()

# Check a entails b
result1 = entails(lambda R, S: evaluate(R, S)[1])

# Check a entails e
result2 = entails(lambda R, S: evaluate(R, S)[4])

print("\nENTAILMENT RESULTS")
print("-" * 40)

print("a entails b:", result1)
print("a entails e:", result2)

if result1:
    print("Therefore, a |= b")
else:
    print("Therefore, a does not entail b")

if result2:
    print("Therefore, a |= e")
else:
    print("Therefore, a does not entail e")
```
