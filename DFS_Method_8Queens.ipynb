```python
import collections
def get_neighbors(state):
    neighbors = []
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)
    moves = [(-1, 0, 'Up'), (1, 0, 'Down'),
             (0, -1, 'Left'), (0, 1, 'Right')]
    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col
            new_state = list(state)
            new_state[zero_idx], new_state[new_idx] = new_state[new_idx], new_state[zero_idx]
            neighbors.append((tuple(new_state), move))
    return neighbors
def dfs(start, goal, max_depth=25):
    stack = collections.deque([(start, [], 0)])
    visited = set()
    while stack:
        state, path, depth = stack.pop()
        if state == goal:
            return path
        if state in visited or depth >= max_depth:
            continue
        visited.add(state)
        for neighbor, move in reversed(get_neighbors(state)):
            if neighbor not in visited:
                stack.append((neighbor, path + [move], depth + 1))
    return None
def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()
def get_input(name):
    print(f"\nEnter {name} (use 0 for blank space):")
    numbers = []
    for i in range(3):
        row = list(map(int, input(f"Row {i+1}: ").split()))
        numbers.extend(row)
    if len(numbers) != 9 or set(numbers) != set(range(9)):
        raise ValueError("Enter numbers 0 to 8 exactly once.")
    return tuple(numbers)
print("=== 8 PUZZLE USING DFS ===")
start = get_input("Initial State")
goal = get_input("Goal State")
print("\nInitial State:")
print_board(start)
print("Goal State:")
print_board(goal)
solution = dfs(start, goal)
if solution is not None:
    print("Solution Found! /n Number of moves:", len(solution))
    print("Moves:", " -> ".join(solution))
    current = start
    print("\nPath to Goal:")
    print_board(current)
    for move in solution:
        for neighbor, direction in get_neighbors(current):
            if direction == move:
                current = neighbor
                print("Move:", move)
                print_board(current)
                break
else:
    print("No solution found within the depth limit.")
```
