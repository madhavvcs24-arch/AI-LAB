```python

def alpha_beta(node, alpha, beta, is_max, path="Root"):

    # Leaf node
    if isinstance(node, int):
        return node

    if is_max:
        best = float('-inf')

        for i, child in enumerate(node):
            value = alpha_beta(
                child, alpha, beta, False,
                path + f"->{i+1}"
            )

            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha and i < len(node) - 1:
                print("Pruned branches at:", path)
                break

        return best

    else:
        best = float('inf')

        for i, child in enumerate(node):
            value = alpha_beta(
                child, alpha, beta, True,
                path + f"->{i+1}"
            )

            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha and i < len(node) - 1:
                print("Pruned branches at:", path)
                break

        return best


# Given game tree
tree = (
    (
        ((5, 6), (7, 4, 5)),
        ((3,),)
    ),
    (
        ((6,), (6, 9)),
        ((7,),)
    ),
    (
        ((5,),),
        ((9, 8), (6,))
    )
)

# Initial alpha and beta
alpha = float('-inf')
beta = float('inf')

# Find optimal value
result = alpha_beta(tree, alpha, beta, True)

print("\nOptimal Root Value =", result)
```
