```python
import time

class VacuumEnvironment:
    def __init__(self):
        # Explicitly set both rooms to DIRTY (1)
        self.locations = {
            'A': 1,
            'B': 1
        }
        # Explicitly set the initial starting position to Room A
        self.vacuum_location = 'A'
        self.cost = 0  # Tracking performance/steps taken

    def display_status(self):
        print(f"Current State -> Room A: {'DIRTY' if self.locations['A'] else 'CLEAN'} | "
              f"Room B: {'DIRTY' if self.locations['B'] else 'CLEAN'}")
        print(f"Vacuum Location: Room {self.vacuum_location}\n" + "-"*40)


class ReflexVacuumAgent:
    def act(self, environment):
        environment.display_status()

        # Goal check: Stop if both rooms are clean
        if environment.locations['A'] == 0 and environment.locations['B'] == 0:
            print("🎉 Success! Both rooms are completely clean.")
            return False

        current_room = environment.vacuum_location
        room_status = environment.locations[current_room]

        # Rule 1: If current room is dirty, clean it
        if room_status == 1:
            print(f"🧹 Action: Room {current_room} is DIRTY. Sucking dirt...")
            time.sleep(1)
            environment.locations[current_room] = 0
            print(f"✨ Room {current_room} is now CLEAN.")

        # Rule 2: If current room is clean, move to the other room
        else:
            if current_room == 'A':
                print("➡️ Action: Room A is clean. Moving RIGHT to Room B.")
                environment.vacuum_location = 'B'
            else:
                print("⬅️ Action: Room B is clean. Moving LEFT to Room A.")
                environment.vacuum_location = 'A'

        environment.cost += 1
        print(f"Accumulated Path Cost: {environment.cost}\n")
        return True


# --- Execution ---
if __name__ == "__main__":
    print("🤖 Initializing Vacuum Cleaner AI Environment...\n")
    env = VacuumEnvironment()
    agent = ReflexVacuumAgent()

    # Run the simulation loop until the environment is fully clean
    running = True
    while running:
        running = agent.act(env)
        time.sleep(1)

    print(f"\nFinal Statistics: Total actions taken = {env.cost}")
```
