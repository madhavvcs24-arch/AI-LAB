```python

import random
import math

N = 8

# Calculate the number of attacking queen pairs
def calculate_cost(board):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


# Generate random initial state
def initialize():
    return [random.randint(0, N - 1) for _ in range(N)]


# Display chessboard
def display(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Simulated Annealing algorithm
def simulated_annealing():

    current = initialize()
    T = 100.0

    while T > 0.001:

        current_cost = calculate_cost(current)

        # Check whether solution is found
        if current_cost == 0:
            break

        # Generate random neighbour
        next_state = current.copy()

        col = random.randint(0, N - 1)
        row = random.randint(0, N - 1)

        next_state[col] = row

        next_cost = calculate_cost(next_state)

        # Calculate change in energy
        deltaE = current_cost - next_cost

        if deltaE > 0:
            current = next_state

        else:
            # Accept worse solution with probability
            p = math.exp(deltaE / T)

            if random.random() < p:
                current = next_state

        # Decrease temperature
        T = T * 0.99

    return current


# Main program
random.seed()

solution = simulated_annealing()

print("Final Chessboard:")
display(solution)

cost = calculate_cost(solution)

print("\nNumber of attacking pairs:", cost)

if cost == 0:
    print("Solution Found!")
else:
    print("Solution Not Found. Try Again!")
```
